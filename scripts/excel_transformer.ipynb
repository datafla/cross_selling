### CONVERSOR DE EXCEL A CSV PARA DATA EXTENSA
Este pequeño pedazo de codigo nos sirve para poder convertir un libro de excel con extensa data distribuida en varias hojas, esto fue creado para obtener un .csv con el que poder trabajar en python

```python
import pandas as pd
import numpy as np
```

### Load the data

```python
# Importamos informacion del arbol de categorias (CUBO VENTAS)
cubo = pd.read_excel(r"C:\Users\fdavila\OneDrive - Farmacorp S.A\Escritorio\cross selling\data\cubo_1.xlsx", 
                    sheet_name = 'SKUs',
                    skiprows=4
                    )
```

```python
# Copiamos la direccion del libro que queremos convertir a .csv
ventas = "C:\\Users\\fdavila\\OneDrive - Farmacorp S.A\\Escritorio\\cross selling\\data\\ventas_ene_2026.xlsx"
```

```python
# Creamos una lista vacia para guardar los DF
dataframes = []

# Definimos la cantidad de hojas que hay en el libro
first = 1
last = 5 

# Definimos nombre del .csv (output)
output = 'data/enero_2026'

# Leemos el rango de hojas del libro
for i in range(first, last):
    nombre_sheet = f'Sheet {i}'
    df = pd.read_excel(ventas, sheet_name=nombre_sheet)
    dataframes.append(df)
    print(f'Hoja leída {nombre_sheet}: {len(df)} filas')

# Concatenamos todos los DF
df_combinado = pd.concat(dataframes, ignore_index=True)

# Exportamos a CSV todas las hojas
df_combinado.to_csv(output, index=False)

print(f'\n✓ Archivo CSV creado con {len(df_combinado)} filas totales')
```
